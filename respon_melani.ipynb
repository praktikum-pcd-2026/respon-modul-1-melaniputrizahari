```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import cv2
```

```python
spider_a = plt.imread('brandnewday.jpg')
plt.imshow(spider_a)
print(spider_a.shape)
```

```python
spider_b = cv2.imread('brandnewday.jpg')
plt.imshow(spider_b)
```

```python
plt.subplot(1,2,1)
plt.imshow(spider_a)
plt.title("gambarA")
plt.axis('on')

plt.subplot(1,2,2)
plt.imshow(spider_b)
plt.title("gambarB")
plt.axis('on')

plt.tight_layout()
plt.show()
```

gambar_a menggunakan plt.imread untuk membaca gambar menggunakan import matplotlib. Hanya membaca gambar aslinya, tidak mengubah warna. Sedangkan gambar_b itu memakai cv2.imread, menggunakan import cv2. import cv2 itu digunakan untuk baca, ubah warna dan manipulasi gambar. Sehingga gambar_b itu menampilkan 3 channel (warna). 

```python
plt.figure(figsize=(6,6))

plt.subplot(2,2,1)
plt.imshow(spider_a[:, :, 0], cmap="Reds")
plt.title("layer1")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,2,2)
plt.imshow(spider_a[:, :, 1], cmap="Greens")
plt.title("layer2")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,2,3)
plt.imshow(spider_a[:, :, 2], cmap="Blues")
plt.title("layer3")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

spider_gray = cv2.cvtColor(spider_a, cv2.COLOR_RGB2GRAY)
plt.subplot(2,2,4)
plt.imshow(spider_gray, cmap="gray")
plt.title("Greyscale")
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.subplot(2,2,1)
plt.imshow(spider_b[:, :, 0], cmap="Blues")
plt.title("layer1")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,2,2)
plt.imshow(spider_b[:, :, 1], cmap="Greens")
plt.title("layer2")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,2,3)
plt.imshow(spider_b[:, :, 2], cmap="Reds")
plt.title("layer3")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

spider_grayb = cv2.cvtColor(spider_b, cv2.COLOR_RGB2GRAY)
plt.subplot(2,2,4)
plt.imshow(spider_grayb, cmap="gray")
plt.title("Greyscale")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.subplot(2,3,1)
plt.imshow(spider_b[:, :, 0], cmap="gray")
plt.title("B1")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,3,2)
plt.imshow(spider_b[:, :, 1], cmap="gray")
plt.title("B2")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,3,3)
plt.imshow(spider_b[:, :, 2], cmap="gray")
plt.title("B3")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,3,4)
plt.imshow(spider_a[:, :, 0], cmap="gray")
plt.title("A1")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,3,5)
plt.imshow(spider_a[:, :, 1], cmap="gray")
plt.title("A2")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(2,3,6)
plt.imshow(spider_a[:, :, 2], cmap="gray")
plt.title("A3")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.subplot(1,3,1)
spider_grayb = spider_grayb.copy()
spider_grayb = (spider_b[:, :, 0] + spider_b[:, :, 1] + spider_b[:, :, 2])/3
plt.imshow(spider_grayb, cmap='gray')
plt.title("Average Method")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(1,3,2)
spider_grayb2 = spider_grayb.copy()
spider_grayb2 = ((spider_b[:, :, 0] * 0.114) + (spider_b[:, :, 1] * 0.587) + (spider_b[:, :, 2] * 0.299)/3)
plt.imshow(spider_grayb2, cmap='gray')
plt.title("OpenCV Manual Method")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

spider_graybb = cv2.cvtColor(spider_b, cv2.COLOR_RGB2GRAY)
plt.subplot(1,3,3)
plt.imshow(spider_graybb, cmap="gray")
plt.title("OpenCV Built In Method")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.subplot(2,2,1)
plt.imshow(spider_graybb, cmap="gray")
spiderb_1 = spider_graybb.copy()
plt.title("Gray")

plt.subplot(2,2,2)
new_spiderb_1 = 255 - spiderb_1
plt.imshow(new_spiderb_1, cmap='gray')
plt.title("Negatif")

plt.subplot(2,2,3)
new_spiderb_2 = spiderb_1 - new_spiderb_1
plt.imshow(new_spiderb_2, cmap='gray')
plt.title("G-N")

plt.subplot(2,2,4)
new_spiderb_3 = new_spiderb_1 - spiderb_1
plt.imshow(new_spiderb_3, cmap='gray')
plt.title("N-G")

plt.tight_layout()
plt.show()
```

```python
def zero_padding(image, pad):
    h, w = image.shape
    padded = np.zeros((h + 2*pad, w + 2*pad), dtype=image.dtype)

    padded[pad:pad+h, pad:pad+w] = image
    return padded

padding = zero_padding(spiderb_1, 150)
plt.imshow(padding, cmap='gray')
```
