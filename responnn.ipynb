Sebagai contoh, area yang seharusnya berwarna merah pada gambar asli akan tampak biru pada gambarB, dan sebaliknya.

Ketiga metode grayscale menghasilkan citra dengan tingkat kecerahan berbeda. Average Method ((R+G+B)/3) menghasilkan citra paling terang karena semua channel diberi bobot sama (1/3), namun kontrasnya lebih rendah dan kurang natural. OpenCV Manual Method (0.299R + 0.587G + 0.114B) menghasilkan citra lebih natural karena meniru sensitivitas mata manusia dengan bobot tertinggi pada channel hijau. OpenCV Built In Method menghasilkan citra yang sama persis dengan Manual Method.

```python
def rotate_image_manual(image, angle):
    h, w = image.shape
    rad = np.radians(angle)
    cos_a, sin_a = np.cos(rad), np.sin(rad)
    
    cx, cy = w // 2, h // 2
    
    y, x = np.indices((h, w))
    
    x_shifted = x - cx
    y_shifted = y - cy
    
    x_orig = x_shifted * cos_a + y_shifted * sin_a + cx
    y_orig = -x_shifted * sin_a + y_shifted * cos_a + cy
    
    x0 = np.floor(x_orig).astype(int)
    x1 = x0 + 1
    y0 = np.floor(y_orig).astype(int)
    y1 = y0 + 1
    
    x0 = np.clip(x0, 0, w-1)
    x1 = np.clip(x1, 0, w-1)
    y0 = np.clip(y0, 0, h-1)
    y1 = np.clip(y1, 0, h-1)
    
    dx = x_orig - x0
    dy = y_orig - y0
    
    Ia = image[y0, x0]
    Ib = image[y1, x0]
    Ic = image[y0, x1]
    Id = image[y1, x1]
    
    result = (1 - dx) * (1 - dy) * Ia + dx * (1 - dy) * Ic + (1 - dx) * dy * Ib + dx * dy * Id
    return result.astype(np.uint8)

sudut = 73

rot_cw = rotate_image_manual(spider_graybb, -sudut)   
rot_ccw = rotate_image_manual(spider_graybb, sudut)  

plt.figure(figsize=(6,6))

plt.subplot(1,2,1)
plt.imshow(rot_cw, cmap='gray')
plt.title("Searah Jarum Jam")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.subplot(1,2,2)
plt.imshow(rot_ccw, cmap='gray')
plt.title("Berlawanan Jarum Jam")
plt.axis('on')
plt.xticks(np.arange(0, 601, 200))
plt.yticks(np.arange(0, 701, 100))

plt.tight_layout()
plt.show()
```
